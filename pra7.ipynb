```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ["A", "B", "C", "D"]
y = [10, 20, 15, 25]

plt.bar(x, y)
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 25]

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 15, 18, 20, 20, 22, 25, 30]

plt.hist(data)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
