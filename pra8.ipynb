```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(2, 1, 1)
plt.plot(x, y)
plt.title("Line Plot")

plt.subplot(2, 1, 2)
plt.bar(x, y)
plt.title("Bar Plot")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y1 = [10, 20, 30]
y2 = [15, 25, 35]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.legend()
plt.title("Multiple Lines")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.title("Grid Example")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, color="red", marker="o", label="Data")
plt.title("Styled Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid()
plt.show()
```
