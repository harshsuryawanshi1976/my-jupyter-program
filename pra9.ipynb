```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

sns.lineplot(x=x, y=y)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = ["A", "B", "C", "D"]
y = [10, 20, 15, 25]

sns.barplot(x=x, y=y)
plt.show()
```
