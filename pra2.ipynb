```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
print(a)
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
```

```python
a = np.arange(1, 10)
print(a)
```

```python
a = np.linspace(1, 10, 5)
print(a)
```

```python
a = np.array([10, 20, 30, 40])
print(a[0])
print(a[2])
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a[0, 1])
print(a[1, 2])
```

```python
a = np.arange(1, 7)
b = np.reshape(a, (2, 3))
print(b)
```

```python


a = np.array([10, 20, 30, 40])
print(np.sum(a))
```

```python
a = np.array([10, 20, 30, 40])
print(np.mean(a))
```

```python
import numpy as np

a = np.arange(1, 7)
b = np.reshape(a, (2, 3))

print("Array:", b)
print("Index:", b[0, 1])
print("Slice:", b[0, :])
print("Sum:", np.sum(b))
print("Mean:", np.mean(b))
```
