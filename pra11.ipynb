```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
B = np.array([5, 6])

X = np.linalg.solve(A, B)
print(X)
```

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])

print(np.linalg.inv(A))
```

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])

values, vectors = np.linalg.eig(A)

print("Eigenvalues:", values)
print("Eigenvectors:")
print(vectors)
```

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
B = np.array([5, 6])

print("Solution:", np.linalg.solve(A, B))
print("Inverse:\n", np.linalg.inv(A))

values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Transpose:\n", A.T)
```
