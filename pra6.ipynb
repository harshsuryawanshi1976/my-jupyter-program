```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Sneha", "Raj"],
    "Marks": [80, None, 90, 70],
    "City": ["Pune", "Mumbai", None, "Pune"]
})

print(df.isnull())
```

```python
df["Marks"] = df["Marks"].fillna(0)

print(df)
```

```python
df2 = df.dropna()

print(df2)
```

```python
result = df[df["Marks"] > 75]

print(result)
```

```python
result = df[["Name", "Marks"]]

print(result)
```

```python
result = df.sort_values("Marks")

print(result)
```

```python
df["City"] = df["City"].replace("Pune", "Delhi")

print(df)
```

```python
result = df.groupby("City")["Marks"].mean()

print(result)
```
