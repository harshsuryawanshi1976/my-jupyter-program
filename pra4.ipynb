```python
import numpy as np

a = np.random.rand(5)
print(a)
```

```python
a = np.random.randn(5)
print(a)
```

```python
a = np.random.randint(1, 10, 5)
print(a)
```

```python
a = np.random.normal(50, 10, 5)
print(a)
```

```python
a = np.random.rand(10)

print("Mean:", np.mean(a))
print("Std:", np.std(a))
```

```python
a = np.random.normal(50, 10, 1000)

print("Mean:", np.mean(a))
print("Std:", np.std(a))
```
